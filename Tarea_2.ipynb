###Ejercicio 1

```python
"""
Simula 10 000 lanzamientos de dos dados y estima la probabilidad
de que la suma sea exactamente 5. Compara esa frecuencia con
la probabilidad clásica que ya calculaste.
"""
import random, matplotlib.pyplot as plt

def lanzar_dado():
  return random.randint(1,6)
n=10000
conteo= 0
frecuencias = []

for i in range(1,n+1):
  dado1 = lanzar_dado()
  dado2 = lanzar_dado()

  if dado1 + dado2 == 5:
    conteo += 1
  frecuencia = conteo/i
  frecuencias.append(frecuencia)

probabilidad = conteo/n
print("Numero de simulaciones:", n)
print("Casos favorables:", conteo)
print("Probabilidad empirica", probabilidad)
print("Probabilidad clasica:", 4/36)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Probabilidad empirica")
plt.axhline(4 / 36, linestyle="--", label="Probabilidad clasica = 4/36")
plt.xlabel("Numero de simulaciones")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad de obtener una suma de 5")
plt.legend()
plt.show()
```

####Ejercicio 2

```python
"""
Simula 10 000 PIN de 4 dígitos, del 0 al 9,
y estima la probabilidad de que ninguno se repita.
Compara el resultado con la fracción de la parte clásica.
"""
import random, matplotlib.pyplot as plt
def generar_pin(digitos):
  return[random.randint(0,9) for _ in range(digitos)]
n=10000
conteo= 0
frecuencias = []

for i in range(1,n+1):
  pin = generar_pin(4)
  if len(set(pin)) == 4:
    conteo += 1
  frecuencias.append(conteo/i)
probabilidad = conteo/n
print("Numero de simulaciones:", n)
print("Pin sin repeticiones:", conteo)
print("Probabilidad empirica", probabilidad)
print("Probabilidad clasica:", 0.504)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Probabilidad empirica")
plt.axhline(0.504, linestyle="--", label="Probabilidad clasica = 0.504")
plt.xlabel("Numero de simulaciones")
plt.ylabel("Frecuencia relativa")
plt.title("PIN de cuatro digitos sin repeticiones")
plt.legend()
plt.show()
```

####Ejercicio 3

```python
"""
Simula 20 000 lanzamientos de tres dados.
Estima la frecuencia de la suma 9 y la de la suma 10.
¿Se nota, en la simulación, cuál de las dos sale más?
"""
import random, matplotlib.pyplot as plt
def lanzar_dado():
  return random.randint(1,6)
n=20000
conteo9=0
conteo10=0
frecuencias9=[]
frecuencias10=[]
for i in range(1,n+1):
  dado1 = lanzar_dado()
  dado2 = lanzar_dado()
  dado3 = lanzar_dado()
  suma = dado1 + dado2 + dado3
  if suma == 9:
    conteo9 += 1
  if suma == 10:
    conteo10 += 1
  frecuencias9.append(conteo9/i)
  frecuencias10.append(conteo10/i)

print("Lanzamientos:", n)
print("Veces que salio 9:", conteo9)
print("Veces que salio 10:", conteo10)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(frecuencias9, label="Suma 9")
plt.plot(frecuencias10, label="Suma 10")
plt.axhline(25 / 216, linestyle="--",label="Teorica suma 9")
plt.axhline(27 / 216, linestyle=":",label="Teorica suma 10")
plt.xlabel("Numero de ensayos")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion de las sumas 9 y 10")
plt.legend()
plt.show()
```

```python
"""
Basado en el Ejemplo 3, simula una urna con 3 bolas rojas
y 2 azules. Realiza 2,000 extracciones (con reemplazo)
y calcula la probabilidad de obtener una azul.
"""
import random, matplotlib.pyplot as plt
def sacar_bolas():
  urna =["roja","roja","roja","azul","azul"]
  return random.choice(urna)
n=2000
conteo_azules=0
frecuencias=[]
for i in range(1,n+1):
  bolas = sacar_bolas()
  if bolas == "azul":
    conteo += 1
  frecuencias.append(conteo/i)
probabilidad = conteo_azules/n
print("Numero en que saco las bolas:", n)
print("Bolas azules obtenidas:", conteo_azules)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Probabilidad empirica")
plt.axhline(2 / 5, linestyle="--",label="Probabilidad clasica = 0.4")
plt.xlabel("Numero de extracciones")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad de obtener una bola azul")
plt.legend()
plt.show()
```

####Ejercicio 5

```python
"""
Una serie son 20 volados. Estima, con 10 000 series,
la probabilidad de que aparezcan 5 águilas seguidas
en algún momento de la serie.
"""
import random, matplotlib.pyplot as plt
def lanzar_moneada():
  return random.choice(["aguila", "sol"])
n=10000
conteo=0
frecuencias=[]
for i in range(1,n+1):
  racha=0
  hubo_racha=False
  for j in range(20):
    resultado=lanzar_moneada()
    if resultado=="aguila":
      racha+=1
      if racha==5:
        hubo_racha=True
        break
    else:
      racha=0
  if hubo_racha:
    conteo+=1
  frecuencias.append(conteo/i)
probabilidad = conteo/n

print("Numero de simulaciones:", n)
print("Series con cinco aguilas seguidas:", conteo)
print("Probabilidad:", probabilidad)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(frecuencias, label="Probabilidad empirica")
plt.xlabel("Numero de series")
plt.ylabel("Frecuencia relativa")
plt.title("Probabilidad de obtener cinco aguilas seguidas")
plt.legend()
plt.show()
```
