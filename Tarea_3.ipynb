##Teorema de bayes

###Ejercicio 1

```python
"""
El 20% de los correos son spam. La palabra «oferta»
aparece en el 80% de los correos spam y en el 10%
de los correos deseados. Si un correo tiene la palabra
«oferta», ¿cuál es la probabilidad de que sea spam?
"""
p_spam = 0.2
p_deseado = 0.8
p_oferta_spam = 0.8
p_oferta_deseado = 0.1

p_spam_oferta = (p_spam * p_oferta_spam) / ((p_spam * p_oferta_spam) + (p_deseado * p_oferta_deseado))

print("Probabilidad de que sea spam:", p_spam_oferta)
print("Porcentaje:", p_spam_oferta * 100, "%")
```

####Ejercicio 2

```python
"""
En una tienda online, el 40% entra por Chrome,
el 35% por Safari y el 25% por otros.
La tasa de conversión (compra) es del 3% en Chrome,
4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable
que haya usado el cliente?
"""
navegadores = {"Chrome": (0.40, 0.03), "Safari": (0.35, 0.04), "Otros": (0.25, 0.02)}

p_compra = 0
for navegador, (p_navegador, p_conversion) in navegadores.items():
    p_compra += p_navegador * p_conversion
print("Probabilidad total de compra:", p_compra)
print("Probabilidades a posteriores por navegador:")
mas_probable = ""
max_p = -1

for navegador, (p_navegador, p_conversion) in navegadores.items():
    p_navegador_compra = (p_navegador * p_conversion) / p_compra
    print(f"{navegador}: {p_navegador_compra * 100:.2f}%")

    if p_navegador_compra > max_p:
        max_p = p_navegador_compra
        mas_probable = navegador

print(f"El navegador más probable que haya usado el cliente es: {mas_probable} ({max_p * 100:.2f}%)")
```

##Probabilidad total

####Ejercicio 1

```python
"""
El 70% de los clientes ven un anuncio y, de ellos, compra el 30%.
 El 30% no ve el anuncio y, de ellos, compra el 10%.
 Calcula la probabilidad total de que un cliente compre.
"""
p_anuncio = 0.7
p_no_anuncio = 0.3
p_compra_anuncio = 0.3
p_compra_no_anuncio = 0.1
p_compra = (p_anuncio * p_compra_anuncio) + (p_no_anuncio * p_compra_no_anuncio)
print("Probabilidad total de compra:", p_compra)
print("Porcentaje:", p_compra * 100, "%")
```

##Probabilidad condicional

###Ejercicio 1

```python
"""
Se lanza un dado equilibrado.
Calcula la probabilidad de obtener un 2,
dado que salió un número par.
"""
espacio = [2, 4, 6]
A = []
for x in espacio:
    if x == 2:
        A.append(x)
probabilidad = len(A) / len(espacio)

print("Probabilidad:", probabilidad)
```

###Ejercicio 2

```python
"""
Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9,
dado que las tres caras son distintas.
"""
from itertools import product
S = list(product(range(1, 7), repeat=3))
distintos = []
for x in S:
  if len(set(x)) == 3:
    distintos.append(x)
suma_9 = []
for x in distintos:
  if sum(x) == 9:
    suma_9.append(x)
probabilidad = len(suma_9) / len(distintos)

print("Casos favorables:", len(suma_9))
print("Casos posibles:", len(distintos))
print("P(Suma 9 | Caras distintas) =", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```
