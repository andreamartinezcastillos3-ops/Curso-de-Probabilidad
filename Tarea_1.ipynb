###Ejercicio 1

```python
"""
Al lanzar dos dados, calcula la probabilidad de que la suma
de sus caras sea exactamente 5
"""
from itertools import product

D1 = list(product(range(1, 7), repeat=2))
print("Espacio Muestral:", D1)
D2 = []
for x in D1:
    if x[0] + x[1] == 5:
        D2.append(x)
print("Casos favorables:", D2)
print("Probabilidad:", len(D2)/len(D1))
```

###Ejercicio 2

```python
"""
En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad
de que ninguno de los dígitos se repita.
"""
from itertools import product

C = list(product(range(10), repeat=3))
c1 = []
for x in C:
    if len(set(x)) == 3:
        c1.append(x)

print("Resultados posibles:", len(C))
print("Casos favorables:", len(c1))
print("Probabilidad:", len(c1)/len(C))
```

###Ejercicio 3

```python
"""
Tienes 8 canciones distintas y las pones en un orden aleatorio,
con todos los órdenes igualmente probables.
Calcula la probabilidad de que tu favorita quede primera
 y la que menos te gusta quede última.
"""
from itertools import permutations

canciones = [1, 2, 3, 4, 5, 6, 7, 8]
casos_posibles = list(permutations(canciones))
casos_favorables = []
for x in casos_posibles:
    if x[0] == 1 and x[-1] == 8:
        casos_favorables.append(x)

print("Resultados posibles:", len(casos_posibles))
print("Casos favorables:", len(casos_favorables))
print("Probabilidad cancion favorita:", len(casos_favorables) / len(casos_posibles))
```

###Ejercicio 4

```python
"""
Se lanzan tres dados equilibrados.
¿Qué es más probable, que la suma sea 9 o que la suma sea 10?
¿Por cuántos resultados le gana una suma a la otra?
"""
from itertools import product

D = list(product(range(1, 7), repeat=3))
suma_9 =[]
suma_10 = []
for x in D:
    if sum(x) == 9:
        suma_9.append(x)
    if sum(x) == 10:
        suma_10.append(x)
diferencia = len(suma_9) - len(suma_10)
print("Casos favorables suma 9:", len(suma_9))
print("Probabilidad:", len(suma_9)/len(D))
print("Casos favorables suma 10:", len(suma_10))
print("Probabilidad:", len(suma_10)/len(D))
print("La diferencia es:", diferencia)
```

###Ejercicio 5

```python
"""
Ana, Bruno, Carla y Diego hacen un amigo secreto:
cada quien saca un regalo al azar
y todos los repartos son igualmente probables.
Calcula la probabilidad de que nadie saque su propio regalo.
"""
from itertools import permutations

amigo_secret = ["Ana", "Bruno", "Carla", "Diego"]
R = list(permutations(amigo_secret))
E = []

for x in R:
    if x[0] != "Ana" and x[1] != "Bruno" and x[2] != "Carla" and x[3] != "Diego":
        E.append(x)

print("Resultados posibles:", len(R))
print("Casos favorables (nadie se autoregala):", len(E))
print("Probabilidad:", len(E)/len(R))
```
